# JavaScript: Last-Minute Revision

A compact recap of the concepts practiced across JavaScript basics, control flow, arrays, functions, closures, objects, asynchronous JavaScript, modules, classes, OOP, TypeScript, Playwright, and coding exercises.

## 1. Values, Variables, and Types

- Primitive values: `string`, `number`, `boolean`, `undefined`, `null`, `bigint`, `symbol`. Objects, arrays, and functions are reference values.
- Prefer `const` by default, `let` when reassignment is needed, and avoid `var` in modern code. `const` prevents rebinding, not mutation of an object.
- Identifiers are case-sensitive, cannot start with a digit, and cannot be reserved words. Use `camelCase` for variables/functions and `PascalCase` for classes.
- `typeof null` is the historical exception: it returns `'object'`. Test null with `value === null`.
- `undefined` often means missing; `null` is intentional emptiness. `NaN` has type `number`; use `Number.isNaN(value)` to check it.
- BigInt uses an `n` suffix (`42n`); do not mix BigInt and Number in arithmetic without converting.

```js
const MAX_RETRIES = 3;
let attempts = 0;
const isReady = false;
```

**Remember:** `const` means constant binding, not deeply immutable data.

## 2. Operators, Coercion, and Decisions

- Arithmetic: `+ - * / % **`; `+` also concatenates strings. Assignment operators such as `+=` update a binding.
- Prefer `===` and `!==`: they compare without type coercion. Relational comparisons of two strings are lexicographic.
- Falsy values: `false`, `0`, `-0`, `0n`, `''`, `null`, `undefined`, `NaN`. Empty arrays and objects are truthy.
- `&&` and `||` short-circuit and return an operand; `!` negates truthiness. Use `??` when only nullish values should trigger a fallback.
- Ternary: `condition ? yes : no`. Use `if / else` when branches need explanation. In `switch`, remember `break` unless fall-through is intentional.

```js
const label = score >= 70 ? 'PASS' : 'RETRY';
const displayName = user.name ?? 'Guest';
```

**Pitfall:** `0 || 10` gives `10`; `0 ?? 10` gives `0`.

## 3. Loops and Input

- `for` fits a known iteration pattern; `while` repeats while a condition is true; `do...while` runs at least once.
- `break` exits the nearest loop/switch; `continue` skips to the next iteration. Ensure each loop makes progress toward stopping.
- `for...of` iterates values; `for...in` iterates enumerable keys and is usually not the right array loop.
- A nested loop over `n` items is commonly O(n^2). Trace outer and inner indices separately.
- Input is environment-specific: prompt libraries, Node `readline`, and file input with `fs` work differently. Convert text explicitly (`Number(input)`) and validate it.

```js
for (let index = 0; index < items.length; index++) {
  console.log(index, items[index]);
}
```

**Trace rule:** record the loop variable before and after each iteration when predicting output.

## 4. Functions

- Parameters are inputs; `return` is the output. `console.log` displays a value but does not return it.
- Function declarations are hoisted. Function expressions and arrow functions are values assigned to bindings.
- Arrow functions capture surrounding `this`; they are not interchangeable with methods that need their own `this`.
- Default parameters apply when an argument is `undefined`. Rest parameters (`...values`) collect arguments into an array.
- Spread (`...array`) expands iterable values. Spread creates a shallow copy, not a deep copy.
- Pure functions depend only on inputs and avoid side effects, making them easier to test.

```js
const add = (left, right = 0) => left + right;
function sumAll(...values) {
  return values.reduce((total, value) => total + value, 0);
}
```

**Remember:** a function without an explicit `return` returns `undefined`.

## 5. Arrays

- Arrays are ordered and zero-indexed. `length` is one past the highest index; arrays can be sparse.
- Read with `array[index]` or `array.at(-1)`. Check with `includes`; search for an item with `find`.
- `push / pop` act at the end; `unshift / shift` at the start. All four mutate.
- `slice(start, end)` copies a range without mutation; end is excluded. `splice(start, count, ...items)` changes the original.
- `map` transforms, `filter` selects, `reduce` accumulates, `forEach` performs an action and returns `undefined`.
- `some` checks whether any item matches; `every` checks whether all match.
- `sort` mutates and sorts as strings by default. Numeric ascending: `(a, b) => a - b`; descending: `(a, b) => b - a`.
- `concat` and spread make shallow copies; nested objects/arrays remain shared references.

```js
const scores = [8, 3, 10];
const doubled = scores.map(score => score * 2);
const ascending = [...scores].sort((a, b) => a - b);
```

**Pitfall:** copy before sorting if the original order must remain unchanged.

## 6. Strings

- Strings are immutable; methods return new strings. Useful tools: `length`, `at`, `includes`, `slice`, `split`, `join`, `trim`, `replace`, `replaceAll`, and case conversion.
- Template literals use backticks and `${expression}`.
- For case-insensitive comparison, normalize both sides first. For palindrome checks, decide how to treat spaces and punctuation.
- Anagram check: normalize, then compare sorted characters or character-frequency maps.

**Pattern:** normalize -> transform -> compare.

## 7. Objects and Multidimensional Arrays

- Objects store key/value properties. Dot access is for known keys; bracket access supports dynamic keys (`user[field]`).
- Add/update by assignment; remove with `delete`. `Object.keys`, `Object.values`, and `Object.entries` inspect properties.
- Destructuring extracts values; rename with `{ name: fullName }`; set defaults with `{ role = 'user' }`.
- Object spread merges shallowly: `{ ...defaults, ...settings }`; later properties win.
- Optional chaining (`user.address?.city`) stops on nullish values; combine with `??` for a fallback.
- A grid is an array of arrays: `grid[row][column]`. Loop over rows, then cells.
- Objects and arrays compare by identity, not contents.

```js
const { name, address: { city } } = user;
const updated = { ...user, active: true };
```

## 8. Scope, Hoisting, and Closures

- `let` and `const` are block-scoped; `var` is function-scoped.
- `var` is initialized to `undefined`; `let` and `const` are unavailable in the temporal dead zone until declaration executes.
- Function declarations can be called before their source line; a `let`/`const` function expression cannot be used before initialization.
- A closure is a function plus access to its surrounding lexical scope, even after the outer function returns. Closures support private state and function factories.

```js
function makeCounter() {
  let count = 0;
  return () => ++count;
}
const next = makeCounter();
```

**Pitfall:** reading a `let`/`const` binding before its declaration throws `ReferenceError`, not `undefined`.

## 9. Callbacks and the Event Loop

- A callback is a function passed to another function to be called later or when work completes.
- `setTimeout` schedules work; it does not block the rest of the script.
- Synchronous code runs first, then Promise handlers (microtasks), then timer callbacks (tasks).
- Deeply nested callbacks make sequencing/error handling difficult; Promises and `async / await` flatten the flow.

```js
console.log('A');
setTimeout(() => console.log('timer'), 0);
Promise.resolve().then(() => console.log('promise'));
console.log('B'); // A, B, promise, timer
```

## 10. Promises

- A Promise is `pending`, then `fulfilled` or `rejected`; settled Promises do not change state again.
- `then` handles fulfillment and returns a new Promise; `catch` handles rejection; `finally` runs cleanup.
- Return the next Promise from a `then` handler to sequence steps. Throwing in a handler rejects the chain.
- `Promise.all` fulfills when all fulfill and rejects on the first rejection. `allSettled` waits for every outcome.
- `race` settles with the first settled input; `any` fulfills with the first fulfillment (rejects if all reject).

```js
fetchData()
  .then(data => transform(data))
  .then(result => console.log(result))
  .catch(error => console.error(error))
  .finally(() => cleanup());
```

**Pitfall:** forgetting `return` inside `then` breaks the intended chain.

## 11. Async / Await

- Calling an `async` function always returns a Promise. Returning a value fulfills it; throwing rejects it.
- `await` pauses that async function, not the JavaScript thread. Handle failures with `try / catch / finally`.
- Await independent operations together with `Promise.all`; await sequentially when one step depends on the previous result.
- `await` inside a loop is sequential; use it for dependent work, not automatically for independent tasks.

```js
async function loadBoth() {
  const [users, teams] = await Promise.all([getUsers(), getTeams()]);
  return { users, teams };
}
```

## 12. Modules

- ES modules split code into files with explicit dependencies using `export` and `import`.
- Named exports use braces and matching names. A default export is imported without braces and can receive a local name.
- Use clear relative paths for local modules and keep exports intentional. Project configuration determines Node module mode; keep it consistent.

```js
export const multiply = (a, b) => a * b;
import { multiply } from './math.js';
```

## 13. Classes and OOP

- A class describes shared behavior; `new` creates an instance and calls `constructor`. Instance fields hold state; methods define behavior; static members belong to the class.
- `this` in an instance method usually refers to the instance when called as `instance.method()`. Detaching the method can lose that receiver.
- **Encapsulation:** keep state and behavior together; `#field` creates a private field.
- **Inheritance:** `extends` reuses/specializes a base class. Call `super(...)` before using `this` in a derived constructor.
- **Polymorphism:** different objects respond to the same method call in their own way. Program against shared behavior.
- **Abstraction:** expose essential operations and hide implementation detail. Prefer composition to deep inheritance when appropriate.

```js
class Page {
  verify() { return 'generic check'; }
}
class LoginPage extends Page {
  verify() { return 'check login controls'; }
}
```

## 14. Playwright

- `test` defines a test; fixtures such as `{ page }` provide a browser page. Navigate with `await page.goto(url)`.
- Await browser actions and assertions. Prefer user-facing locators: `getByRole`, `getByLabel`, `getByText`, or a stable test id.
- Locators auto-wait for actionability. Web-first assertions such as `await expect(locator).toBeVisible()` retry until success or timeout.
- Test behavior, not implementation details. Keep tests isolated and locators specific.
- Configuration can set test directory, browser projects, retries, workers, reporter, base URL, and tracing.

```ts
import { test, expect } from '@playwright/test';
test('opens the page', async ({ page }) => {
  await page.goto('/');
  await expect(page.getByRole('heading', { name: 'Home' })).toBeVisible();
});
```

## 15. TypeScript Essentials

- TypeScript adds static checks and compiles to JavaScript. Types catch many mistakes before runtime; they do not validate external data automatically.
- Common types: `string`, `number`, `boolean`, `T[]`, object types. Inference often makes local annotations unnecessary.
- `type` and `interface` describe shapes; `?` marks an optional property; `|` forms a union; `&` combines types.
- Narrow unions with `typeof`, `in`, or a discriminant before using type-specific properties.
- Generics preserve input/output relationships. Avoid `any`; use `unknown` for untrusted values and narrow it.

```ts
type Status = 'idle' | 'loading' | 'done';
interface User { name: string; email?: string; }
function first<T>(items: T[]): T | undefined { return items[0]; }
```

**Remember:** compile-time types guide code; runtime checks protect real inputs.

## 16. Coding-Exercise Patterns

- **Sum/count:** initialize an accumulator and scan once: O(n) time, O(1) extra space.
- **Min/max:** start with the first item, compare the rest, and define behavior for empty input.
- **Frequency/duplicates:** use a `Map` or object count; usually O(n) time and O(n) space.
- **Reverse:** two pointers and swap inward, or copy then reverse if mutation is unwanted.
- **Second min/max:** track best and next-best; decide whether duplicates count.
- **Palindrome:** normalize and compare inward, or compare with a reversed copy.
- **Anagram:** normalize and compare character counts; sorting both strings is simpler but O(n log n).
- **Sorting:** use a numeric comparator. Bubble sort teaches nested loops (O(n^2)); practical code usually uses built-in `sort`.
- **Patterns/grids:** outer loop controls rows; inner loop controls columns or printed symbols.

**Before coding:** clarify inputs, output, empty cases, duplicates, mutation, and expected complexity.

## 17. Final Quick-Recall Checklist

- Use `===`; know the falsy values; empty arrays/objects are truthy.
- `const` prevents rebinding, not mutation. `let`/`const` are block-scoped; `var` is function-scoped.
- `map` transforms, `filter` selects, `reduce` accumulates, `forEach` performs side effects.
- `slice` copies; `splice` mutates. `sort` mutates and needs a numeric comparator.
- Objects/arrays compare by identity. Spread copies only one level.
- A closure remembers lexical scope. An async function returns a Promise.
- Use `Promise.all` for independent work; await dependent steps in order.
- In Playwright, await actions and use auto-retrying, user-facing assertions.
- In TypeScript, narrow `unknown`; don't mistake static types for runtime validation.
